```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

def vif(x: pd.DataFrame(), threshold: float = 5):
  while True:
    vif_data = pd.DataFrame()
    vif_values = []
    vif_data['Features'] = x.columns

    for i in range(len(x.columns)):
      vif = variance_inflation_factor(x.values, i)
      vif_values.append(vif)

    vif_data['VIF values'] = vif_values

    if vif_data['VIF values'].max() > threshold:
      max_index = vif_data['VIF values'].idxmax()
      coln = vif_data['Features'].loc[max_index]
      x.drop(coln, axis = 1, inplace= True)

    else: break

  return vif_data
```
